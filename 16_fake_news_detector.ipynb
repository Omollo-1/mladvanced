# 16/ Fake News Detector

**Dataset (Kaggle):** [https://www.kaggle.com/datasets/clmentbisaillon/fake-and-real-news-dataset](https://www.kaggle.com/datasets/clmentbisaillon/fake-and-real-news-dataset)

> Contains two CSVs: Fake.csv and True.csv, each with title, text, subject and date columns.

**Goal:** Classify a news article as FAKE or REAL based on its text.

**How to get the data:**
1. Download from the link above (or run `kaggle datasets download -d <dataset-slug> -p data/ --unzip`, requires a `kaggle.json` API token in `~/.kaggle/`).
2. Place the extracted file(s) in a local `data/` folder next to this notebook.
3. Update `DATA_PATH` in the cell below if your filename differs.


```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

sns.set_style("whitegrid")
```

```python
fake = pd.read_csv("data/Fake.csv")
real = pd.read_csv("data/True.csv")
fake['label'] = 'FAKE'
real['label'] = 'REAL'
df = pd.concat([fake, real], ignore_index=True).sample(frac=1, random_state=42).reset_index(drop=True)
df['content'] = df['title'].astype(str) + " " + df['text'].astype(str)
print(df.shape)
df.head()
```

## Exploratory Data Analysis

```python
print(df['label'].value_counts())
sns.countplot(y='label', data=df, order=df['label'].value_counts().index)
plt.title("Label distribution")
plt.show()

df['text_length'] = df['content'].astype(str).apply(len)
df['text_length'].hist(bins=50)
plt.title("Text length distribution")
plt.show()
```

## Text Cleaning

```python
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\S+", " ", text)
    text = re.sub(r"<.*?>", " ", text)
    text = re.sub(r"[^a-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df['clean_text'] = df['content'].apply(clean_text)
df[['content', 'clean_text']].head()
```

## Train/Test Split & TF-IDF Vectorization

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['clean_text'], df['label'], test_size=0.2, random_state=42, stratify=df['label']
)

tfidf = TfidfVectorizer(max_features=20000, ngram_range=(1,2), stop_words='english')
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
print(X_train_tfidf.shape)
```

## Model Training

```python
model = LogisticRegression(max_iter=1000, class_weight='balanced')
model.fit(X_train_tfidf, y_train)
```

## Evaluation

```python
y_pred = model.predict(X_test_tfidf)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted"); plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

## Try It Yourself

```python
def predict_sample(text):
    cleaned = clean_text(text)
    vec = tfidf.transform([cleaned])
    return model.predict(vec)[0]

predict_sample("Replace this string with your own example text")
```
